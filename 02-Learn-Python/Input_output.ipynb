## Reading CSV File


```python
import csv

with open('/content/sample_data/california_housing_test.csv', mode = 'r') as file:
  csv_reader = csv.reader(file)
  for row in csv_reader:
    print(row)
```

```python
import csv

data = [
    ['Name', 'Age', 'Country'],
    ['Alice', 25, 'USA'],
    ['Bob', 32, 'Canada']
    ]

with open('output.csv', mode = 'w') as file:
  csv_writer = csv.writer(file)
  csv_writer.writerows(data)
```

## JSON File in Python
1. Store data in key-value pairs
2. Separated module to read JSON file

```python
import json

with open('/content/sample_data/anscombe.json', mode = 'r') as file:
  data = json.load(file)
  print(data)
```

```python
# to write json
import json

data = {'Name': 'Alice', 'Age': 25, 'country': 'USA'}

with open('output2.json', mode = 'w') as file:
  json.dump(data, file)
```

# Exercise
You're given a string.
Your task: First write the input string to a file 'example.txt'. Read the same file and return string.

```python
# Function to write and read an input text file
def read_file(text):

  with open('example.txt', mode = 'w') as file:
    write_file = file.write(text)

  with open('/content/example.txt', mode = 'r') as file:
    read_file = file.read()
    return read_file
```

```python
text = "I am blessed, so blessed"
read_file(text)
```
