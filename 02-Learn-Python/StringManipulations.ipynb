# Strings in Python
1. Sequence of character
2. They can be letters, numbers, symbols
3. Example: "Hello World"

```python
greet = "Hello Data Engineering"
print(greet)
```

```python
name = ' John!'
Message = greet + name
print(Message)
```

```python
new_string = greet.replace("Hello", 'Hi')
print(new_string)
```

Removing leading and trelling spaces us strip()

```python
sentence = '       Hello World       '
print(sentence)
print(sentence.strip())
```

```python
index = greet.find('Data')
print(index)
```

```python
template = 'hello {}'
print(template.format('Daniel'))
```

```python
print(greet.upper())
```

```python
print(greet.lower())
```

```python
words = greet.split()
print(words)
```
